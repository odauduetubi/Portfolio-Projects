## Data Loading

Now, since we have our data, we load it as follows:

````python
import pandas as pd

df1 = pd.read_excel("data/online_retail_II.xlsx", sheet_name="Year 2009-2010")
df2 = pd.read_excel("data/online_retail_II.xlsx", sheet_name="Year 2010-2011")
````

````python
df1.shape, df2.shape
````

Now, we join the datasets together.

````python
df = pd.concat([df1, df2])
````

````python
df = df.reset_index(drop=True)
df
````

````python
df.columns = df.columns.str.lower().str.replace(' ', '_')
````
